## 1. Problem

> Given an array of integers `nums` and an integer `target`, return the *indices*
> of the two numbers that add up to `target`. Exactly one solution exists, and you
> may not use the same element twice.

**Example 1**
```
Input:  nums = [2, 7, 11, 15], target = 9
Output: [0, 1]
```

**Constraints**
- `2 <= len(nums) <= 10^4`
- Exactly one valid answer exists

## 2. Thinking

- **Brute force idea:** check every pair → $O(n^2)$.
- **Key insight:** for each `x`, I need `target - x`. A dict of *value → index* of
  numbers seen so far answers "have I seen it?" in $O(1)$.
- **Edge cases to remember:** duplicates like `[3, 3]`, negative numbers.

````python
# Setup: run this cell first. Lets you run pytest tests inside the notebook.
import ipytest
import pytest

ipytest.autoconfig()
````

## 3. Solution

### Approach A — brute force
**Complexity:** Time $O(n^2)$ · Space $O(1)$

````python
def two_sum_brute(nums: list[int], target: int) -> list[int]:
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return [i, j]
    return []
````

### Approach B — one-pass hash map
**Complexity:** Time $O(n)$ · Space $O(n)$

````python
def two_sum(nums: list[int], target: int) -> list[int]:
    seen = {}  # value -> index
    for i, x in enumerate(nums):
        if target - x in seen:
            return [seen[target - x], i]
        seen[x] = i
    return []
````

## 4. Tests

Both implementations are tested against the same cases by parametrizing over the function too.

````python
%%ipytest -q

CASES = [
    ([2, 7, 11, 15], 9, [0, 1]),
    ([3, 2, 4], 6, [1, 2]),
    ([3, 3], 6, [0, 1]),          # duplicates
    ([-1, -2, -3, -4], -7, [2, 3]),  # negatives
]

@pytest.mark.parametrize("fn", [two_sum_brute, two_sum])
@pytest.mark.parametrize("nums, target, expected", CASES)
def test_two_sum(fn, nums, target, expected):
    assert fn(nums, target) == expected
````

## 6. Takeaways

- **Pattern:** "find a complement" → store what you've seen in a dict.
- **Mistakes I made:** inserting `x` into `seen` *before* checking, which matched an element with itself.
- **Related problems:** 3Sum, Two Sum II (sorted input → two pointers).
